```python
#1 & 2)
seconds_per_minute = 60
minutes_per_hour = 60
seconds_per_hour = seconds_per_minute * minutes_per_hour
print(f"{seconds_per_hour} seconds are in an hour")

#3 & 4)
hours_per_day = 24
seconds_per_day = seconds_per_hour * hours_per_day
print(f"{seconds_per_day} seconds are in a full day")

#5)
float_division = seconds_per_day / seconds_per_hour
print(float_division)

#6)
int_division = seconds_per_day // seconds_per_hour
print(int_division)
#Yes the integer and float numbers agreed aside from the .0
```
